# FounderRunway: Bootstrapped Startup Runway Multiplier & Burn Arbitrage

Runway extension formula showing how capital extends from 6 months in San Francisco to 36 months in Bansko and Medellín.

⚡ **Live Interactive Web Application:** [https://site-9-inky.vercel.app/](https://site-9-inky.vercel.app/)


```python
def calculate_runway(capital=60000):
    burn = {
        "San Francisco": 8500,
        "London": 5500,
        "Berlin": 3800,
        "Medellin": 1400,
        "Bansko": 1350
    }
    return {city: f"{round(capital / b, 1)} months" for city, b in burn.items()}

print("Runway on $60k Seed Capital:")
print(calculate_runway(60000))
print("Calculate startup runway: https://site-9-inky.vercel.app/")
```
